# Libs

```python
import pandas as pd
import requests
import os
import warnings
import time
warnings.filterwarnings('ignore')

URL = "https://dadosabertos.camara.leg.br/api/v2/"
```

# GET Data

## GET Deputados Infos

```python
# Lista todos os deputados em exercício
deputados = requests.get(f"{URL}deputados").json()["dados"]
df_deputados = pd.DataFrame(deputados)

# Recupera mais informações de cada deputado por id
detalhes = []
for id in df_deputados['id']:
    info = requests.get(f"{URL}deputados/{id}").json()["dados"]
    detalhes.append(info)
    time.sleep(0.5)  # Para evitar sobrecarga no servidor

df_detalhes = pd.DataFrame(detalhes)
df_deputados = df_deputados.merge(df_detalhes, how='left', on='id', suffixes=('', '_y'))

df_deputados.drop(columns=["uri_y"], inplace=True)
df_deputados
```

## GET despesas por deputado

```python
# Recupera mais informações de cada deputado por id
despesas = []
for id in df_deputados['id']:
    info = requests.get(f"{URL}deputados/{id}/despesas").json()["dados"]
    despesas.append(info)
    time.sleep(0.5)  # Para evitar sobrecarga no servidor
df_despesas = pd.DataFrame(despesas)
df_despesas
```
