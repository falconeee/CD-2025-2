# Trabalho Prático

O objetivo geral deste trabalho é praticar os conceitos discutidos em sala de aula, principalmente: representação e pré-processamento de dados textuais; redução de dimensionalidade e algoritmos de agrupamento.

## Conjunto de Dados

Para esse trabalho utilizaremos um conjunto de dados de jogos da Steam. Os dados foram obtidos de datasets públicos da Steam disponibilizados no Kaggle.

O conjunto de dados que vocês deverão usar é o arquivo `steam_games.csv` (já processado e disponível no diretório do trabalho). Os dados estão organizados em um arquivo (`.csv`) com as seguintes colunas:

* **genres** - gêneros a que um jogo pertence. Veja que um jogo pode estar associado a mais de um gênero (`str`);
* **description** - descrição detalhada do jogo (`str`);
* **release_date** - data de lançamento do jogo (`str`);
* **name** - nome/título do jogo (`str`);
* **average_playtime** - tempo médio de jogo, em minutos (`int`);
* **positive_ratings** - número de avaliações positivas (`int`);
* **negative_ratings** - número de avaliações negativas (`int`);
* **price** - preço do jogo (`float`);
* **appid** - ID único do jogo na Steam (`int`);
* **short_description** - descrição resumida do jogo (`str`).

**Observação:** esse conjunto de dados foi processado a partir dos datasets originais da Steam.

**ATENÇÃO**
É possível que seu computador não consiga lidar com o conjunto de dados completo. Se for o caso, faça uma amostra aleatória de jogos do conjunto de dados. Deixe claro o processo de amostragem, o tamanho das amostras finais que considerou no trabalho e as especificações do computador utilizado.

Os dados para o trabalho estão disponíveis no AVA da disciplina, Tópico "Trabalho Prático".

## Objetivo

Vocês deverão utilizar aprendizado não supervisionado (neste caso, agrupamento) para investigar se há relação entre a **descrição** de um jogo e o(s) **gênero(s)** ao(s) qual(quais) ele pertence.

De forma mais específica, vocês deverão agrupar os jogos de acordo suas descrições e, após isso, verificar a distribuição dos gêneros em cada grupo. Ou seja, se os jogos pertencentes ao mesmo grupo possuem o(s) mesmo(s) gênero(s) ou se as distribuições de gêneros de grupos diferentes são distintas.

Obs: Você pode utilizar tanto a descrição detalhada (`description`) quanto a descrição resumida (`short_description`) para realizar os agrupamentos. Se achar interessante, também pode incluir o título da obra. Justifique sua escolha.

Obs 2: Você pode desconsiderar gêneros pouco frequentes (ex: Accounting) ou labels que não são gêneros (ex: Early Access, Free to Play). Justifique sua escolha.

## Metodologia

Para atingir o objetivo proposto, vocês deverão seguir ao menos os seguintes passos:

1. Pré-processamento dos dados textuais;
2. Construção da matriz de TF-IDF;
3. Redução de dimensionalidade, via PCA (Leia também sobre o *Truncated SVD*: https://scikit-learn.org/stable/modules/generated/sklearn.decomposition.TruncatedSVD.html);
4. Aplicação de ao menos dois algoritmos de agrupamento vistos na disciplina;
5. Validação da metodologia utilizada;
6. Interpretação dos resultados.

**PS:** essas são as exigências mínimas. Caso vocês queiram fazer/propor algo extra, será permitido (e pode ajudar na nota se a proposta for relevante).

### Abordagem com LLM (Extra)

O intuito dessa etapa é explorar o uso de [Modelos de Linguagem](https://arxiv.org/abs/1706.03762) para representar as descrições dos jogos a partir de [embeddings](https://www.ibm.com/br-pt/think/topics/embedding), capazes de capturar relações semânticas dos textos, ao invés da tradicional abordagem baseada em TF-IDF, focada na frequência dos termos.

**Importante**: Essa etapa é opcional e valerá até 1 ponto extra na nota final do trabalho.

Passos sugeridos:

1. Escolha um modelo pequeno da biblioteca [`SentenceTransformers`](https://www.sbert.net/) (eg. `all-MiniLM-L6-v2`, `paraphrase-MiniLM-L6-v2`)
2. Gere embeddings para as descrições dos jogos
3. **Importante**: Avalie se a estratégia de agrupamento precisa ser alterada quando usando embeddings ao invés de TF-IDF. Se sim, tente entender e justificar o porquê.
4. **Agrupamento**: Aplique ao menos um algoritmo de agrupamento (pode ser diferente dos vistos em aula)
5. **Classificação por Similaridade**: Implemente uma abordagem alternativa usando similaridade de cosseno:
   - Gere embeddings para os rótulos (gêneros) disponíveis
   - Para cada jogo, calcule a similaridade de cosseno entre o embedding da descrição e os embeddings dos gêneros
   - Atribua cada jogo ao(s) gênero(s) mais similar(es)
   - Compare os resultados com os gêneros reais dos jogos

**Análise Comparativa**: Compare os resultados obtidos com sentence embeddings versus TF-IDF. Qual abordagem gerou agrupamentos mais coerentes com os gêneros reais?
