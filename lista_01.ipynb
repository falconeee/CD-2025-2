### Exemplo

Vamos começar com um exemplo sobre a distribuição Uniforme contínua.

A biblioteca ``scipy`` possui funcionalidades para gerar números pseudo-aleatórios e para computar valores da CDF teórica da distribuição. Leia mais sobre em https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.uniform.html.

A biblioteca ``statsmodels`` possui funcionalidades para computar a CDF empírica de um conjunto de observações.

```python
from scipy.stats import uniform
import matplotlib.pyplot as plt
from statsmodels.distributions.empirical_distribution import ECDF
```

```python
def plot_uniform_cdf(loc, scale, n):
  # gera 'n' números de uma distribuição uniforme contínua no
  # intervalo [loc, loc + scale]
  r = uniform.rvs(loc = loc, scale = scale, size = n)

  # plota a CDF empírica dos números gerados
  cdf = ECDF(r)
  plt.plot(cdf.x, cdf.y, linewidth = 3, alpha = 0.5, color = 'blue',
           label = 'Empírica')

  # plota a CDF teórica de uma distribuição contínua uniforme no intervalo
  # [loc, loc + scale]
  plt.plot(cdf.x, uniform.cdf(cdf.x, loc = loc, scale = scale),
           alpha = 0.5, label = 'Teórica', color = 'red',
           linewidth = 3, zorder = 0);

  # Ajustes de rótulos e tamanho da figura...
  plt.legend(loc = 'best');
  plt.xlabel(r'$x$');
  plt.ylabel(r'CDF');
  plt.gcf().set_size_inches(3.5, 3.5)
```

```python
plot_uniform_cdf(5, 5, 1000)
```

```python
plot_uniform_cdf(0, 1, 1000)
```

**ATENÇÃO:** Leia com atenção a documentação da biblioteca mencionada para entender como o parâmetro $\lambda$ deve ser fornecido. O $\lambda$ da definição acima não é o mesmo que o `scale` da biblioteca.

```python
from scipy.stats import expon
```

```python
def plot_exponential_cdf(lambda_, n):
  # gera 'n' números de uma distribuição exponencial com parâmetro 'lambda_'
  r = expon.rvs(scale = 1/lambda_, size = n)

  # plota a CDF empírica dos números gerados
  cdf = ECDF(r)
  plt.plot(cdf.x, cdf.y, linewidth = 3, alpha = 0.5, color = 'blue', label = 'Empírica')

  # plota a CDF teórica de uma distribuição exponencial com parâmetro 'lambda_'
  plt.plot(cdf.x, expon.cdf(cdf.x, scale = 1/lambda_),
           alpha = 0.5, label = 'Teórica', color = 'red',
           linewidth = 3, zorder = 0);

  # Ajustes de rótulos e tamanho da figura...
  plt.legend(loc = 'best');
  plt.xlabel(r'$x$');
  plt.ylabel(r'CDF');
  plt.gcf().set_size_inches(3.5, 3.5)
```

Repita a análise inicial para a distribuição de pareto. Lembre-se, dado um parâmetro $b > 0$, a densidade da distribuição de pareto é:
$$f(x) = \frac{b}{x^{b+1}}\text{, } x \ge 1.$$

Consulte https://docs.scipy.org/doc/scipy/tutorial/stats/continuous_pareto.html para gerar números pseudo-aleatórios e para computar a CDF da distribuição de pareto.

Você deve:
1. Completar a função abaixo para gerar gráficos comparativos entre as CDFs empíricas e teóricas da distribuição de pareto;
2. Executar seu código, variando os parâmetros da distribuição e número de elementos gerados.

```python
from scipy.stats import pareto
```

```python
def plot_pareto_cdf(b, n):
  # gera 'n' números de uma distribuição de Pareto com parâmetro 'b'
  r = pareto.rvs(b, size=n)

  # plota a CDF empírica dos números gerados
  cdf = ECDF(r)
  plt.plot(cdf.x, cdf.y, linewidth=3, alpha=0.5, color='blue', label='Empírica')

  # plota a CDF teórica de uma distribuição de Pareto com parâmetro 'b'
  plt.plot(cdf.x, pareto.cdf(cdf.x, b),
           alpha=0.5, label='Teórica', color='red',
           linewidth=3, zorder=0);

  # Ajustes de rótulos e tamanho da figura...
  plt.legend(loc='best');
  plt.xlabel(r'$x$');
  plt.ylabel(r'CDF');
  plt.gcf().set_size_inches(3.5, 3.5)
```

Após fazer as figuras, você provavelmente observou que para valores pequenos de $b$, a CDF da distribuição de pareto é bastante difícil de ler, uma vez que fica praticamente "colada" às retas $x = 0$ e $y = 1$.

Esse fenômeno ocorre porque, quando $b$ é pequeno, valores muito grandes (i.e., discrepantes ou *outliers*) têm uma probabilidade não negligível de ocorrerem.

Quando observamos o comportamento das figuras acima, devemos modificar a forma de visualizar a distribuição:
1. Devemos gerar uma figura para a CCDF (*Complementary Cumulative Distribution Function*). A CCDF de um valor $x$ é definida como 1 menos a CDF de $x$.
2. Devemos colocar os eixos $x$ e $y$ em escala logarítmica.

Complete a função abaixo para gerar a figura da forma descrita acima e gere os gráficos para os mesmos casos que testou anteriormente.

```python
def plot_pareto_ccdf(b, n):
  # gera 'n' números de uma distribuição de Pareto com parâmetro 'b'
  r = pareto.rvs(b, size=n)

  # plota a CCDF empírica dos números gerados
  cdf = ECDF(r)
  plt.plot(cdf.x, 1-cdf.y, linewidth=3, alpha=0.5, color='blue', label='Empírica')

  # plota a CCDF teórica de uma distribuição de Pareto com parâmetro 'b'
  plt.plot(cdf.x, 1-pareto.cdf(cdf.x, b),
           alpha=0.5, label='Teórica', color='red',
           linewidth=3, zorder=0);

  # Ajustes de rótulos e tamanho da figura...
  plt.legend(loc='best');
  plt.xlabel(r'$x$');
  plt.ylabel(r'CCDF');
  plt.xscale('log') # escala logarítmica no eixo x
  plt.yscale('log') # escala logarítmica no eixo y
  plt.gcf().set_size_inches(3.5, 3.5)
```

```python
plot_pareto_ccdf(3, 100)
```

```python
plot_pareto_ccdf(3, 1000)
```

```python
plot_pareto_ccdf(2, 1000)
```

```python
plot_pareto_ccdf(1, 1000)
```

```python
plot_pareto_ccdf(0.5, 1000)
```

### Q2:

O artigo introduz o conceito de leis de potência, que descrevem quantidades cuja probabilidade de ocorrência é inversamente proporcional a uma potência de seu valor. Essas distribuições são encontradas em diversas áreas, como biologia, economia, ciências sociais e física, aplicando-se a fenômenos como o tamanho de cidades, a magnitude de terremotos e até mesmo fortuna de pessoas.

Diferente de distribuições com uma escala típica e um valor médio bem definido (como a altura de seres humanos), as quantidades que seguem uma lei de potência podem variar em várias ordens de magnitude. Uma característica dessas distribuições é que, quando representadas em um histograma com eixos logarítmicos (log-log) como nos exemplos utilizados na lista, elas formam uma linha reta. Isso ocorre porque a equação da lei de potência, $p(x)=Cx−α$, quando convertida para uma escala logarítmica se torna uma equaçáo do tipo $\ln{(x)}=\ln{C}−α\ln{x}$, que é a equação de uma reta com inclinação $−α$.

O artigo também fala sobre os desafios e as metodologias para identificar e quantificar corretamente o comportamento de leis de potência em conjuntos de dados empíricos. A abordagem padrão para verificar a existência de uma lei de potência é plotar um histograma dos dados em escalas logarítmicas e observar se o resultado é uma linha reta. Porém o autor demonstra que esse método é inadequado na maioria dos casos.

Utilizando um conjunto de dados gerado a partir de uma lei de potência com expoente $α=2.5$, o artigo mostra que um histograma com bins de tamanho fixo, quando plotado em escala log-log, apresenta um ruído significativo na cauda da distribuição, que é a região de valores mais altos. Isso ocorre porque a frequência de grandes valores é muito baixa, resultando em poucas amostras por bin. Consequentemente, as flutuações estatísticas se tornam grandes em comparação com o número de amostras, o que torna a reta característica difícil de ser identificada com precisão.

Para mitigar o problema do ruído, o artigo apresenta duas soluções principais:

- **Agrupamento Logarítmico**: Uma forma de aprimorar o histograma é utilizar caixas de larguras variáveis. No agrupamento logarítmico, a largura de cada caixa sucessiva é um múltiplo fixo da anterior. Ao fazer isso as caixas na cauda da distribuição se tornam mais largas, capturando mais amostras e reduzindo o erro estatístico nessa região. Ele também destaca que é importante que a contagem em cada bin seja normalizada, dividindo pela largura da caixa, para que a altura represente a densidade de probabilidade.

- **Função de Distribuição Cumulativa**: Este é apresentado como um método superior. Em vez de plotar a frequência dos valores, se plota a probabilidade acumulada P(x) de que uma variável tenha um valor maior ou igual a $x$. Se a distribuição de probabilidade $p(x)$ segue uma lei de potência $p(x)=Cx−α$, então sua forma cumulativa $P(x)$ também seguirá uma lei de potência, mas com um expoente $α−1$. A principal vantagem desse método é que ele elimina a necessidade de agrupar os dados, evitando assim a perda de informação e as ambiguidades relacionadas ao tamanho das caixas.

O autor descreve também como estimar o expoente α e adverte que o método mais comum que é o ajuste de uma linha reta por mínimos quadrados ao gráfico log-log é inadequado, pois é conhecido por introduzir vieses sistemáticos nos resultados. No exemplo do artigo, esse método resultou em um expoente de $α=2.26±0.02$, que é diferente do valor real de 2.5 a partir do qual os dados foram gerados. O autor expõe um método alternativo para o cálculo do α que é o de máxima verossimilhança. Para dados contínuos que seguem uma lei de potência para valores $x≥x_{min}$, o expoente α pode ser estimado com a fórmula:
        $$
        a=1+n[\sum_{i=1}^{n}\ln{\frac{x_i}{x_{min}}}]^{-1}
        $$

Por fim a Função de Distribuição Cumulativa Complementar (CCDF) de uma distribuição de Pareto tem a forma de uma linha reta em um gráfico com eixos logarítmicos porque a própria CCDF também é uma função de lei de potência. Isso acontece porque a CCDF, $P(x)$, é a integral da distribuição de probabilidade da lei de potência, $p(x)=Cx−α$. A integração resulta em uma nova função de lei de potência,   
        $$
        P(x)=\frac{C}{a-1}{x}^{-(a-1)}
        $$
 Como qualquer função de lei de potência se torna linear em uma escala logarítmica, a CCDF também aparece como uma linha reta.

#### 1. O que é o coeficiente de correlação?

É uma medida estatística que avalia o grau de correlação entre duas variáveis, também conhecida como coeficiente de correlação de Pearson.
O valor desse coeficiente varia de -1 a 1, onde:
- -1 indica uma correlação perfeitamente inversa, ou seja, quando uma variável aumenta a outra diminui na mesma proporção;
- +1 indica uma correlação perfeitamente direta, significando que quando uma variável aumenta, a outra também aumenta na mesma proporção;
- 0 indica ausência de correlação linear entre as variáveis.

#### 2. Quando o coeficiente de correlação é útil e quando ele pode ser a ferramenta errada para analisar a relação entre duas variáveis?

O coeficiente de correlação é útil para avaliar o grau, a direção e a força da relação linear entre duas variáveis quantitativas, permitindo identificar se elas variam juntas (correlação positiva) ou em sentidos opostos (correlação negativa). Além disso, o coeficiente de determinação ($r^2$) indica quanto da variação de uma variável pode ser explicada pela outra.

Entretanto, ele pode ser a ferramenta errada quando:
- a relação entre as variáveis é não linear;
- há outliers que distorcem o resultado;
- existem subgrupos com comportamentos distintos;
- os dados não são independentes;
- há heteroscedasticidade (variância desigual);
- as variáveis estão em escala ordinal, caso em que é mais adequado usar a correlação de Spearman.

Logo o coeficiente de correlação se mostra apropriado apenas para relações lineares entre variáveis quantitativas independentes e sem outliers relevantes.

#### 3. Qual a diferença entre causalidade e correlação?

A correlação indica que duas variáveis se relacionam (mudam juntas) mas não necessariamente que uma causa a outra. Já a causalidade representa uma relação de causa e efeito, em que uma variável provoca diretamente a mudança na outra.

Por exemplo, o artigo indica que países com maior consumo de chocolate têm mais ganhadores do Prêmio Nobel e esse é um exemplo de correlação mas essa relação é apenas uma coincidência estatística o chocolate em si não causa inteligência. Outros fatores, como investimento em educação e pesquisa, explicam o fenômeno.

### Parte (a)

Para cada um dos 13 conjuntos de dados disponível no arquivo, calule e imprima:
1. A média dos valores das colunas `x` e `y`
2. O desvio padrão dos valores das colunas `x` e `y`

Com base nos valores observados, formule uma hipótese sobre o grau de similaridade entre os diversos conjuntos de dados. Em outras palavras, vocês diriam que eles são similares ou diferentes?

```python
dataset_groups = df_dados_q5.groupby('dataset')
```

#### 1. Média dos valores X e Y

```python
df_groups_x_mean = pd.DataFrame(dataset_groups.x.mean())
df_groups_y_mean = pd.DataFrame(dataset_groups.y.mean())
df_mean = pd.concat([df_groups_x_mean, df_groups_y_mean],axis=1)
df_mean.rename(columns={'x':'Média X', 'y':'Média Y'}, inplace=True)
df_mean
```

#### 2. Desvio padrão dos valores X e Y

```python
df_groups_x_std = pd.DataFrame(dataset_groups.x.std())
df_groups_y_std = pd.DataFrame(dataset_groups.y.std())
df_std = pd.concat([df_groups_x_std, df_groups_y_std],axis=1)
df_std.rename(columns={'x':'Desvio Padrão X', 'y':'Desvio Padrão Y'}, inplace=True)
df_std
```

#### 3. Correlação linear entre X e Y

```python
df_corr = dataset_groups.apply(lambda g: g['x'].corr(g['y']))
df_corr = pd.DataFrame(df_corr, columns=['Correlacao entre X e Y'])
df_corr
```

### Parte (b)

Agora, vamos testar a hipótese estabelecida no item anterior. Faça um gráfico, para cada um dos 13 conjuntos de dados, marcando os pontos de cada conjunto no plano.

Com base nos gráficos, o que pode ser dito sobre o grau de similaridade entre os diversos conjuntos de dados? Em outras palavras, vocês diriam que eles são similares ou diferentes?

```python
for dataset_name, group in dataset_groups:
    ax = group.plot(
        x='x', y='y',
        kind='scatter',
        title=f'Dispersão entre X e Y — Dataset: {dataset_name}',
        figsize=(5, 5),
        sharex=False,
        sharey=False
    )
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    plt.show()
```
